# Update User Subscription Plans
This notebook allows you to easily assign or update subscription plans for users based on their email address.

```python
import os
import sys
from dotenv import load_dotenv

# Ensure the backend src directory is in the path
sys.path.append(os.path.abspath(os.path.join('..', 'src')))

# Load environment variables
load_dotenv(os.path.join('..', '.env'))

from fastapi_app.utils.supabase_client import get_supabase_admin_client
client = get_supabase_admin_client()
print("Supabase client initialized.")
```

```python
# Fetch available subscription plans
res = client.table("subscription_plans").select("id, tier_name").execute()
plans = res.data
plan_map = {plan['tier_name']: plan['id'] for plan in plans}

print("Available Plans:")
for name, pid in plan_map.items():
    print(f"- {name}: {pid}")
```

## Define Target Subscriptions
Define a dictionary mapping user emails to their desired subscription plan.

```python
user_subscriptions = {
    "contactkrishnababu@gmail.com": "Growth",
    "moorthyk66@gmail.com": "Starter",
    "nivedass321@gmail.com": "Pro",
    "rkselvakumar@gmail.com": "Pro",
    "kishore.rakesh53@gmail.com": "Starter"
}
```

```python
for email, tier_name in user_subscriptions.items():
    if tier_name not in plan_map:
        print(f"❌ Unknown tier '{tier_name}' for {email}")
        continue
        
    plan_id = plan_map[tier_name]
    
    # Update the profile
    res = (
        client.table("profiles")
        .update({"subscription_tier": plan_id})
        .eq("email", email)
        .execute()
    )
    
    if res.data:
        print(f"✅ Successfully updated {email} to {tier_name}")
    else:
        print(f"⚠️  Could not find or update user {email}")
```
